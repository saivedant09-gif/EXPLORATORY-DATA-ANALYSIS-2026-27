```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
mnist_df = pd.read_csv('/content/mnist_test.csv')
```

### Data Understanding: Number of Samples, Classes, and Image Dimensions

```python

X = mnist_df.drop('label', axis=1)
y = mnist_df['label']

num_samples = X.shape[0]
print(f'Number of samples: {num_samples}')

num_classes = y.nunique()
print(f'Number of classes: {num_classes}')

image_dim = int(np.sqrt(X.shape[1]))
print(f'Image dimensions: {image_dim}x{image_dim} pixels')
```

### Visualize Sample Handwritten Digit Images

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 6))
axes = axes.flatten()

for i in range(10):
    idx = np.random.randint(0, num_samples)

    image_data = X.iloc[idx].values.reshape(image_dim, image_dim)

    axes[i].imshow(image_data, cmap='gray')
    axes[i].set_title(f'Label: {y.iloc[idx]}')
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```

### Normalize Pixel Values to the Range [0,1]

```python
X_normalized = X / 255.0

print('Pixel values normalized to the range [0,1].')
print('First 5 rows of normalized data:')
display(X_normalized.head())
```
