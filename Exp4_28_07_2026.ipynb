***HANDLING MISSING VALUES***

```python
import pandas as pd
import numpy as np
```

***Creating Student Performance Dataset***

```python
data = {
    "Math": [88, 76, 95, 67, 84],
    "Science": [91, 80, 89, 72, 85],
    "English": [85, 78, 90, 70, 82]
}

df = pd.DataFrame(
    data,
    index=["Rahul", "Priya", "Amit", "Sneha", "Karan"]
)

print("Original DataFrame")
print(df)
```

***Adding Missing Values***

```python
df["Computer"] = np.nan

df.loc["Anjali"] = [90, 92, 88, 95]

df.loc["Rohan"] = np.nan

df["Sports"] = np.nan

df.loc["Rahul", "Computer"] = 98

print("\nDataFrame with Missing Values")
print(df)
```

***Checking Missing Values***

```python
print("\nIs Null")
print(df.isnull())

print("\nNot Null")
print(df.notnull())
```

***Dropping Missing Values***

```python
print("\nComputer Column without NaN")
print(df["Computer"].dropna())

print("\nDrop rows containing any NaN")
print(df.dropna())

print("\nDrop rows where all values are NaN")
print(df.dropna(how="all"))

print("\nDrop columns where all values are NaN")
print(df.dropna(how="all", axis=1))

print("\nDrop columns having less than 5 non-NaN values")
print(df.dropna(thresh=5, axis=1))
```

***Mathematical Operations with NaN***

```python
arr = np.array([50, 60, np.nan, 80])

series = pd.Series(arr)

print("\nNumPy Mean")
print(arr.mean())

print("\nPandas Mean")
print(series.mean())

computer_marks = df["Computer"]

print("\nSum of Computer Marks")
print(computer_marks.sum())

print("\nAverage Computer Marks")
print(computer_marks.mean())

print("\nCumulative Sum")
print(computer_marks.cumsum())
```

***Filling Missing Values***

```python
filled_df = df.fillna(0)

print("\nDataFrame after fillna(0)")
print(filled_df)

print("\nMean Before Filling")
print(df.mean(numeric_only=True))

print("\nMean After Filling")
print(filled_df.mean(numeric_only=True))
```

***Forward Filling***

```python
print("\nForward Fill (Computer)")
print(df["Computer"].fillna(method="ffill"))
```

***Backward Filling***

```python
print("\nBackward Fill (Computer)")
print(df["Computer"].fillna(method="bfill"))
```

***IMPUTATION TECHNIQUES***

```python
import numpy as np
import pandas as pd

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import SimpleImputer, IterativeImputer
```

***Create Employee Datase***

```python
df = pd.DataFrame(
    {
        "Experience": [2, 5, np.nan, 8, 4],
        "Salary": [30000, np.nan, 45000, 70000, 50000],
        "Department": ["HR", "IT", "Finance", np.nan, "IT"],
    }
)

print("Original DataFrame")
print(df)
```

***Mean Imputation (Numerical)***

```python
mean_imputer = SimpleImputer(strategy="mean")

df["Experience_Mean"] = mean_imputer.fit_transform(df[["Experience"]])
```

***Median Imputation (Numerical)***

```python
median_imputer = SimpleImputer(strategy="median")

df["Salary_Median"] = median_imputer.fit_transform(df[["Salary"]])
```

***Mode Imputation (Categorical)***

```python
mode_imputer = SimpleImputer(strategy="most_frequent")

df["Department_Mode"] = mode_imputer.fit_transform(df[["Department"]]).ravel()
```

***MICE Imputation***

```python
num_cols = ["Experience", "Salary"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = df[num_cols].copy()
df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)
df["Experience_MICE"] = df_mice["Experience"]
df["Salary_MICE"] = df_mice["Salary"]
print("\nDataFrame after Imputation")
print(df)
```
