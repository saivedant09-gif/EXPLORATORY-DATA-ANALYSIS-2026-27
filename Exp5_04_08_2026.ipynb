***TASK-1: Discretization and binning***

```python
import pandas as pd
import numpy as np
```

```python
simple_data = [10, 15, 22, 28, 35, 41, 48, 55, 62, 68, 75, 81, 88, 95]
```

```python
simple_bins = [0, 25, 50, 75, 100]
simple_labels = ['Low', 'Medium', 'High', 'Very High']
simple_categories_cut = pd.cut(simple_data, bins=simple_bins, labels=simple_labels, include_lowest=True)
display(simple_categories_cut)
display(pd.Series(simple_categories_cut).value_counts())
```

```python
simple_q_labels = ['Bottom 33%', 'Middle 33%', 'Top 33%']
simple_categories_qcut = pd.qcut(simple_data, q=3, labels=simple_q_labels, duplicates='drop')
display(simple_categories_qcut)
display(pd.Series(simple_categories_qcut).value_counts())
```

***TASK-2: Outlier detection and filtering***

```python

df_fraud_test = pd.read_csv('/content/fraudTest.csv')
display(df_fraud_test.head())
display(df_fraud_test.info())
```

```python

Q1_amt = df_fraud_test['amt'].quantile(0.25)
Q3_amt = df_fraud_test['amt'].quantile(0.75)
IQR_amt = Q3_amt - Q1_amt
lower_bound_amt = Q1_amt - 1.5 * IQR_amt
upper_bound_amt = Q3_amt + 1.5 * IQR_amt

print(f"Q1 (amount): {Q1_amt:.2f}")
print(f"Q3 (amount): {Q3_amt:.2f}")
print(f"IQR (amount): {IQR_amt:.2f}")
print(f"Lower Bound (amount): {lower_bound_amt:.2f}")
print(f"Upper Bound (amount): {upper_bound_amt:.2f}")
```

```python

outliers_amt = df_fraud_test[(df_fraud_test['amt'] < lower_bound_amt) | (df_fraud_test['amt'] > upper_bound_amt)]

print(f"Number of outliers detected: {len(outliers_amt)}")
print("Sample of outliers (amount):")
display(outliers_amt[['amt', 'is_fraud']].head())
```

```python

df_fraud_test_filtered = df_fraud_test[~((df_fraud_test['amt'] < lower_bound_amt) | (df_fraud_test['amt'] > upper_bound_amt))]

print(f"Original dataset size: {len(df_fraud_test)}")
print(f"Dataset size after filtering outliers: {len(df_fraud_test_filtered)}")
display(df_fraud_test_filtered.head())
```

***TASK-3: Feature scaling and normalization***



```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
df = pd.read_csv('/content/fraudTrain.csv')

print("Original Data (fraudTrain.csv head):\n", df[['amt']].head(), "\n")
```

### Min-Max Scaling (Normalization)

```python
min_max_scaler = MinMaxScaler()
df_min_max_amt = pd.DataFrame(
 min_max_scaler.fit_transform(df[['amt']]),
 columns=['amt']
)
print("Min-Max Scaled Data for 'amt' column:\n", df_min_max_amt.head(), "\n")
```

### Z-score Standardization

```python
standard_scaler = StandardScaler()
df_standard_amt = pd.DataFrame(
 standard_scaler.fit_transform(df[['amt']]),
 columns=['amt']
)
print("Standardized Data for 'amt' column:\n", df_standard_amt.head(), "\n")
```

### Robust Scaling

```python
robust_scaler = RobustScaler()
df_robust_amt = pd.DataFrame(
 robust_scaler.fit_transform(df[['amt']]),
 columns=['amt']
)
print("Robust Scaled Data for 'amt' column:\n", df_robust_amt.head(), "\n")
```
