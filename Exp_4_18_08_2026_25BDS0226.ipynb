```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/WineQT.csv")
```

```python
print(df.head())
print(df.tail(3))
print(df.dtypes)
print(df.describe())
```

```python
print(df.isna().any())
print(df.isnull().sum())
```

```python
print(df[df['quality'].isna()])
```

```python
print(df.isnull().sum().sum())
```

```python
print(df.columns[df.isnull().sum()>0])
```

```python
print(round(df['alcohol'].mean(),2))
print(df['alcohol'].median())
print(df['alcohol'].mode())
```

```python
df['quality'].value_counts().nlargest(5).plot(kind='bar')
plt.show()
```

```python
sns.histplot(df['alcohol'],kde=True)
plt.show()
```

```python
sns.boxplot(df['alcohol'])
plt.show()
```

```python
plt.scatter(df['alcohol'],df['quality'])
plt.title('Alcohol vs Quality')
plt.xlabel('Alcohol')
plt.ylabel('Quality')
plt.show()
```

```python
df['Alcohol_level']=np.where(df['alcohol']>=10,'high','low')
print(df)
```

```python
plt.title('Alcohol Level')
sns.boxplot(x='Alcohol_level',y='alcohol',data=df)
plt.show()
```

```python
sns.pairplot(df[['alcohol','quality','pH','Alcohol_level']],
             hue='Alcohol_level'
             )
plt.show()
```

```python
print(df['alcohol'].corr(df['quality']))
```

```python
print(df.corr(numeric_only=True))
```
